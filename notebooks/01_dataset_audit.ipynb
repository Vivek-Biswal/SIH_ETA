```python
import pandas as pd
import json
from pathlib import Path

RAW = Path("../data/raw")

edges = pd.read_csv(RAW / "IRN_edges.csv")
delays = pd.read_csv(RAW / "train_routes_delays_Sep2024.csv")
routes = pd.read_csv(RAW / "train_routes_Sep2024.csv")

with open(RAW / "stations_zones_mapping.json", "r", encoding="utf-8") as f:
    station_zones = json.load(f)

print("✅ All datasets loaded successfully!")
```

```python
# ==========================================
# STEP 3: JOURNEY RECONSTRUCTION
# ==========================================

# Make copies so the raw data remains untouched
d = delays.copy()
r = routes.copy()

# Convert train/date identifiers to consistent types
d["train"] = d["train"].astype(str)
r["trainNumber"] = r["trainNumber"].astype(str)
d["date"] = d["date"].astype(str)

# Sort route stops in journey order
r = r.sort_values(["trainNumber", "stnSerialNumber"])

# Create station sequence lookup
route_sequence = r[
    ["trainNumber", "station_code", "stnSerialNumber", "distance"]
].drop_duplicates()

# Merge sequence information into delay observations
journey = d.merge(
    route_sequence,
    left_on=["train", "station"],
    right_on=["trainNumber", "station_code"],
    how="left"
)

print("Journey table created.")
print("Rows:", len(journey))
print("Columns:", journey.columns.tolist())
```

```python
print("========== JOURNEY RECONSTRUCTION CHECK ==========")

print("Rows with missing station sequence:",
      journey["stnSerialNumber"].isna().sum())

print("Rows with missing distance:",
      journey["distance"].isna().sum())

print("\nSample reconstructed data:")
display(
    journey[
        [
            "train",
            "date",
            "station",
            "stnSerialNumber",
            "distance",
            "arr_delay",
            "dep_delay"
        ]
    ].head(15)
)
```

```python
# Check how many unique stops each train has
stops_per_train = (
    journey.groupby("train")["stnSerialNumber"]
    .nunique()
)

print("========== ROUTE LENGTH ==========")
print(stops_per_train.describe())

print("\nShortest routes:")
print(stops_per_train.nsmallest(10))

print("\nLongest routes:")
print(stops_per_train.nlargest(10))
```

```python
# ==========================================
# STEP 5: INVESTIGATE ROUTE MISMATCHES
# ==========================================

problem_trains = ["13212", "20978", "20952", "4718"]

for train_id in problem_trains:
    print("\n" + "=" * 60)
    print("TRAIN:", train_id)

    train_route = routes[
        routes["trainNumber"].astype(str) == train_id
    ].sort_values("stnSerialNumber")

    print("Route stations:")
    display(
        train_route[
            [
                "stnSerialNumber",
                "station_code",
                "station_name",
                "distance",
                "arrivalTime",
                "departureTime"
            ]
        ]
    )
```

```python
# Compare the observed stations against the route stations
for train_id in problem_trains:
    observed = set(
        delays.loc[
            delays["train"].astype(str) == train_id,
            "station"
        ].unique()
    )

    route_stations = set(
        routes.loc[
            routes["trainNumber"].astype(str) == train_id,
            "station_code"
        ].unique()
    )

    print("\n" + "=" * 60)
    print("TRAIN:", train_id)
    print("Observed but NOT in route:", sorted(observed - route_stations))
    print("Route but NOT observed:", sorted(route_stations - observed))
```

```python
# ==========================================
# STEP 6: RECONSTRUCT ACTUAL DATETIMES
# ==========================================

import pandas as pd

# Work on a copy
ts = journey.copy()

# Convert service date
ts["service_date"] = pd.to_datetime(ts["date"], errors="coerce")

# Convert time strings to time-of-day
ts["actual_arr_time"] = pd.to_datetime(
    ts["act_arr"],
    format="%I:%M %p",
    errors="coerce"
).dt.time

ts["actual_dep_time"] = pd.to_datetime(
    ts["act_dep"],
    format="%I:%M %p",
    errors="coerce"
).dt.time

ts["scheduled_arr_time"] = pd.to_datetime(
    ts["sch_arr"],
    format="%I:%M %p",
    errors="coerce"
).dt.time

ts["scheduled_dep_time"] = pd.to_datetime(
    ts["sch_dep"],
    format="%I:%M %p",
    errors="coerce"
).dt.time

print("Time conversion complete.")
```

```python
# ==========================================
# STEP 7: HANDLE MIDNIGHT CROSSINGS
# ==========================================

def build_timestamp(date_series, time_series):
    return pd.to_datetime(
        date_series.astype(str) + " " + time_series.astype(str),
        errors="coerce"
    )

ts["actual_arr_dt"] = build_timestamp(
    ts["service_date"],
    ts["actual_arr_time"]
)

ts["actual_dep_dt"] = build_timestamp(
    ts["service_date"],
    ts["actual_dep_time"]
)

ts["scheduled_arr_dt"] = build_timestamp(
    ts["service_date"],
    ts["scheduled_arr_time"]
)

ts["scheduled_dep_dt"] = build_timestamp(
    ts["service_date"],
    ts["scheduled_dep_time"]
)
```

```python
# ==========================================
# STEP 8: SHIFT TIMES AFTER MIDNIGHT
# ==========================================

def fix_midnight(group):
    group = group.sort_values("stnSerialNumber").copy()

    time_columns = [
        "actual_arr_dt",
        "actual_dep_dt",
        "scheduled_arr_dt",
        "scheduled_dep_dt"
    ]

    for col in time_columns:
        # Move timestamp forward by one day whenever it
        # would otherwise move backwards.
        values = group[col].tolist()

        for i in range(1, len(values)):
            if pd.notna(values[i]) and pd.notna(values[i-1]):
                while values[i] < values[i-1]:
                    values[i] += pd.Timedelta(days=1)

        group[col] = values

    return group


ts = (
    ts.groupby(["train", "date"], group_keys=False)
      .apply(fix_midnight)
      .reset_index(drop=True)
)

print("Midnight correction complete.")
```
